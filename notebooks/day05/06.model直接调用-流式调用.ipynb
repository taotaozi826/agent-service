```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 1. 加载环境变量
load_dotenv()  # 加载本地的.env文件,将文件中的内容合并到环境变量

# 获取环境变量中的base_url和api key
dashscope_api_key = os.getenv('DASHSCOPE_API_KEY')
dashscope_base_url = os.getenv('DASHSCOPE_BASE_URL')

# 2. 初始化模型(自定义初始化)
# langchain会根据模型名称自动推断模型厂商决定base_url ,从环境变量中自动读取对应的api_key
model = init_chat_model(
    model='qwen3.8-max',
    model_provider='openai',
    base_url=dashscope_base_url,
    api_key=dashscope_api_key,
    temperature=1.0,
    top_p=0.75,
    # 额外参数,在跟模型交互的过程中会将额外的参数合并到请求体中一起发送给模型
    extra_body={
        "thinking": {
            "type": "disabled"
        }
    }
)

# 4. 直接调用模型(流式调用:模型生成完一部分内容之后就返回, 返回的是一个个片段)
response = model.stream([
    {"role": "system", "content": "你扮演小团团，用她的身份回答问题"},
    {"role": "user", "content": "你是谁？"}
])

for chunk in response:
    print(chunk.content,end='',flush=True)
```
