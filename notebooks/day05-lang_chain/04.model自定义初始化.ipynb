```python
import os

from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

# 1. 加载环境变量
load_dotenv()  # 加载本地的.env文件,将文件中的内容合并到环境变量

# 获取环境变量中的base_url和api key
dashscope_api_key = os.getenv('DASHSCOPE_API_KEY')
dashscope_base_url = os.getenv('DASHSCOPE_BASE_URL')


# 2. 初始化模型(自定义初始化)
# langchain框架不支持的模型需要指定model_provider='openai',base_url 和 api_key
model = init_chat_model(
    model='qwen3.8-max',
    model_provider='openai',
    base_url = dashscope_base_url,
    api_key = dashscope_api_key,
    temperature=1.5,
    top_p=0.75,
    # 额外参数,在跟模型交互的过程中会将额外的参数合并到请求体中一起发送给模型
    extra_body={
        "thinking": {
            "type": "disabled"
        }
    }
)

tools = []

# 3. 创建Agent(绑定模型和工具)
agent = create_agent(model=model, tools=tools)

# 4. 调用模型
response = agent.invoke({'messages': [{'role': 'user', 'content': '你是谁?'}]})

# 5. 输出模型的返回结果
print(response)
```
